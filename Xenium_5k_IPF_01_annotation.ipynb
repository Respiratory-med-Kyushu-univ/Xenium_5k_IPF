```python
coords =sub_adata.obsm["spatial"]

di = 80

all_subtypes = sub_adata.obs["subtype"].unique()
all_subtypes = sorted(all_subtypes) 

score_list = []
```

```python
if os.path.isfile(f"{path}temp_data/temp/smc_{str(di)}.csv"):
    print("already done!")
    score_df = pd.read_csv(f"{path}temp_data/temp/smc_{str(di)}.csv", index_col=0)
else:
    for idx in tqdm.tqdm(range(len(smc_adata))):
        # 中心細胞の座標を取得
        target_coord = smc_adata.obsm["spatial"][idx]
    
        # 距離計算
        distances = np.linalg.norm(coords - target_coord, axis=1)
        
        # 周囲150μm以内の細胞
        mask = (distances <= di) 
        neighbor_indices = np.where(mask)[0]
    
        neighbor_distances = distances[neighbor_indices]
        neighbor_subtypes = sub_adata.obs["subtype"].values[neighbor_indices]
        
        # Gaussian weights
        # weights = np.exp(- (neighbor_distances ** 2) / (2 * sigma ** 2))
        weights = 1
        
        # subtypeごとにスコア集計
        weighted_df = pd.DataFrame({
            "subtype": neighbor_subtypes,
            "weight": weights
        })
        subtype_score = weighted_df.groupby("subtype")["weight"].sum()
    
        # 全 subtype に対するスコア（欠けているものは0に）
        full_score = {sub: 0.0 for sub in all_subtypes}
        full_score.update(subtype_score.to_dict())
    
        score_list.append(full_score)
    # 最終的な DataFrame（上皮細胞数 × subtype 数）
    score_df = pd.DataFrame(score_list, index=smc_adata.obs_names)
    score_df.to_csv(f"{path}temp_data/temp/smc_{str(di)}.csv")
```
